## C03E1 — Campaña, retención y soporte con datos completos

### Situación completa

Una empresa de servicios digitales planifica la próxima semana de trabajo del área Experiencia de Clientes. La gerente decide cuántas horas de calendario asignar a tres actividades. Campaña C son horas dedicadas a preparar audiencias, contenidos y envíos para contactar potenciales clientes. Retención R son horas destinadas a analizar señales de abandono y ejecutar contactos u ofertas para clientes actuales en riesgo. Soporte S son horas de una ventana reforzada de atención para resolver incidentes de clientes activos. Cada hora programada aporta, respectivamente, 9, 13 y 8 puntos de valor esperado. Consume 4, 6 y 3 unidades del presupuesto operativo, cuyo máximo semanal es 120. Además requiere 2, 3 y 4 horas-persona del equipo por cada hora de actividad programada, porque en cada frente trabajan simultáneamente equipos de distinto tamaño; hay 80 horas-persona disponibles. ## C03E3 — Regla dura o faltante penalizado

### Situación completa

La gerente de Experiencia de Clientes revisa la política de retención de la próxima semana. En la formulación base, C son horas de campaña, R horas de retención y S horas de soporte; sus aportes son 9, 13 y 8 puntos por hora. Usan presupuesto según 4, 6 y 3 unidades por hora, con máximo 120, y equipo según 2, 3 y 4 horas-persona, con máximo 80. Se mantienen C entre 0 y 20, R entre 5 y 15 y S entre 4 y 12. La dirección considera dos políticas distintas. La primera convierte R>=10 en una regla obligatoria. La segunda conserva el mínimo base R>=5, fija una meta de 10 y permite declarar un faltante u>=0 tal que R+u>=10; cada hora faltante resta 20 puntos. Así, u mide explícitamente cuánto se incumple la meta flexible, no horas de trabajo ejecutadas. La decisión debe mostrar por separado aporte operativo, penalización y valor formal. Importa no confundir ambos modelos: uno prohíbe R<10; el otro lo permite y expresa cuánto valor asigna la organización a evitar ese faltante.

### Pregunta de decisión

¿Qué plan produce cada política y qué diferencia operativa existe entre exigir diez horas de retención y permitir faltante con una penalización de 20 puntos por hora?

### Vocabulario operativo

- **Regla obligatoria:** Condición que todo plan factible debe cumplir, sin posibilidad de compensación.
- **Meta flexible:** Nivel deseado cuyo incumplimiento se representa y valora dentro del modelo.
- **Faltante u:** Horas necesarias para completar la meta de diez cuando R queda por debajo.
- **Penalización:** Pérdida de 20 puntos por unidad de faltante, aprobada como preferencia organizacional.
- **Valor neto:** Aporte operativo de C, R y S menos la penalización por u.

### Datos y significado

- 10 horas es el nuevo mínimo obligatorio del modelo duro y la meta del flexible.
- 5 horas sigue siendo el piso absoluto de R en el modelo flexible.
- u se mide en horas de retención no cubiertas y satisface R+u>=10.
- 20 puntos por hora es la penalización aplicada a u.
- 9, 13 y 8 son aportes por hora de C, R y S antes de penalizar.
- 120 unidades de presupuesto y 80 horas-persona siguen limitando ambos modelos.
- C<=20, R<=15 y 4<=S<=12 conservan las cotas del caso base.

### Formulación a desarrollar en clase

Modelo flexible: agregar u>=0, recuperar 5<=R<=15 e imponer R+u>=10, equivalente a -R-u<=-10. Maximizar 9C+13R+8S-20u bajo las mismas capacidades, donde u no consume presupuesto ni equipo. Resolver ambos por separado y reportar C,R,S, aporte operativo 9C+13R+8S, u, penalización 20u y valor neto. Cada ejecución conserva recursos, cotas y unidades para que la comparación sea válida.
