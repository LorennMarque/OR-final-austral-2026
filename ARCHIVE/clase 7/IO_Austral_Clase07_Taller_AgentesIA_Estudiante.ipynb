```python
nodes = [1, 2, 3, 4]

variables = [
    ['1','2', 1],
    ['2','1', 1],
    ['3','4', 1],
    ['4','3', 1],
]

# Iniciar solver
solver = pywraplp.Solver.CreateSolver("GLOP")

# Variables
x = {}
for var in variables:
    arco = f"{var[0]}-{var[1]}"
    x[arco] = solver.IntVar(0, 1, arco)


# RESTRICCIONES
# Flujos entrantes = 1
for nodo in nodes:
    solver.Add(
        sum(
            x[f"{var[0]}-{var[1]}"]
            for var in variables
            if var[1] == str(nodo)
        ) == 1
    )

# Flujos salientes = 1
for nodo in nodes:
    solver.Add(
        sum(
            x[f"{var[0]}-{var[1]}"]
            for var in variables
            if var[0] == str(nodo)
        ) == 1
    )


# Función objetivo
# Minimizar costo total de los pesos de los arcos

# Resolver
status = solver.Solve()

print("Estado de la solución:", status)
# 0 Optimo
# 1 Feasible
```
