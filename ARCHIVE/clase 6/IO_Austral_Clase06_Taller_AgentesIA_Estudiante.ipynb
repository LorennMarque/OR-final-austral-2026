```python
# Pegue aquí únicamente la implementación obtenida después de acordar el modelo.
# Debe imprimir estado, variables, objetivo recomputado y residuos.

trimestres = ['Q1', 'Q2', 'Q3', 'Q4']

veleros_iniciales = 10
demanda = [40 , 60, 75, 25]

prod_maxima_trimestre = [40, 40, 40, 40]

costo_prod_normal = 400 # usd

costo_prod_extra = 450 # usd

precio_conserva_siguiente_trimestre = 20 # usd

# ¿Cuánto producir por cada modalidad y cuánto almacenar para minimizar el costo total sin atrasar demanda?
```

```python
from ortools.linear_solver import pywraplp

solver = pywraplp.Solver.CreateSolver("GLOP")

# Variables
# ----------------------------------------------------------------------------

# Veleros producidos en horario regular en Qt, Veleros producidos en horario extra en Qt, Inventario al finalizar Qt
# Produccion entre 0 y 40
# Produccion extra mayor o igual a 0
# El Restante debe ser mayor o igual a 0

# TRIMETRE 1 - VARIABLES
prod_regular_Q1 = solver.NumVar(lb=10, ub=40, name="prod_regular_Q1")
prod_extra_Q1 = solver.NumVar(lb=0, ub=solver.infinity(), name="prod_extra_Q1")
inventario_Q1 = solver.NumVar(lb=0, ub=solver.infinity(), name="inventario_Q1")

# TRIMETRE 2 - VARIABLES
prod_regular_Q2 = solver.NumVar(lb=0, ub=40, name="prod_regular_Q2")
prod_extra_Q2 = solver.NumVar(lb=0, ub=solver.infinity(), name="prod_extra_Q2")
inventario_Q2 = solver.NumVar(lb=0, ub=solver.infinity(), name="inventario_Q2")

# TRIMETRE 3 - VARIABLES
prod_regular_Q3 = solver.NumVar(lb=0, ub=40, name="prod_regular_Q3")
prod_extra_Q3 = solver.NumVar(lb=0, ub=solver.infinity(), name="prod_extra_Q3")
inventario_Q3 = solver.NumVar(lb=0, ub=solver.infinity(), name="inventario_Q3")

# TRIMETRE 4 - VARIABLES
prod_regular_Q4 = solver.NumVar(lb=0, ub=40, name="prod_regular_Q4")
prod_extra_Q4 = solver.NumVar(lb=0, ub=solver.infinity(), name="prod_extra_Q4")
inventario_Q4 = solver.NumVar(lb=0, ub=solver.infinity(), name="inventario_Q4")

# ----------------------------------------------------------------------------
# Restricciones

# Stock que llega = Stock que sale => Stock que tenemos + produccion = demanda + stock que sobra
solver.Add(inventario_Q1 == 10 + prod_regular_Q1 + prod_extra_Q1 - demanda[0])
solver.Add(inventario_Q2 == inventario_Q1 + prod_regular_Q2 + prod_extra_Q2 - demanda[1])
solver.Add(inventario_Q3 == inventario_Q2 + prod_regular_Q3 + prod_extra_Q3 - demanda[2])
solver.Add(inventario_Q4 == inventario_Q3 + prod_regular_Q4 + prod_extra_Q4 - demanda[3])

# ----------------------------------------------------------------------------
# Funcion objetivo

# Min SUMATORIA POR TRIMESTRE (400 * Regular + 450 * Extra + 20 * Inventario)
solver.Minimize(
    400 * prod_regular_Q1 + 450 * prod_extra_Q1 + 20 * inventario_Q1 +
    400 * prod_regular_Q2 + 450 * prod_extra_Q2 + 20 * inventario_Q2 +
    400 * prod_regular_Q3 + 450 * prod_extra_Q3 + 20 * inventario_Q3 +
    400 * prod_regular_Q4 + 450 * prod_extra_Q4 + 20 * inventario_Q4
)
```

```python
# ----------------------------------------------------------------------------
# Estado y resultado del solver
import pandas as pd

status = solver.Solve()
if status == pywraplp.Solver.OPTIMAL:
    print("Solución óptima encontrada.")
    print(f"Costo mínimo total: {solver.Objective().Value():.2f}")

    # Crear un DataFrame con los valores de las variables por trimestre
    resultados = pd.DataFrame({
        "prod_regular": [
            prod_regular_Q1.solution_value(),
            prod_regular_Q2.solution_value(),
            prod_regular_Q3.solution_value(),
            prod_regular_Q4.solution_value()
        ],
        "prod_extra": [
            prod_extra_Q1.solution_value(),
            prod_extra_Q2.solution_value(),
            prod_extra_Q3.solution_value(),
            prod_extra_Q4.solution_value()
        ],
        "inventario": [
            inventario_Q1.solution_value(),
            inventario_Q2.solution_value(),
            inventario_Q3.solution_value(),
            inventario_Q4.solution_value()
        ]
    }, index=["Q1", "Q2", "Q3", "Q4"])
    display(resultados)
else:
    print("No se encontró solución óptima.")
```

## 3. Auditoría obligatoria

1. Reconstruya el objetivo sin usar el valor informado por el solver.
2. Compruebe cada balance o regla.

```python
def validar_solucion(solucion_df, costo_inventario=20, limite_extra=None, inventario_inicial=10):
    """
    Auditoría de la solución:
    1. Re-cálculo del costo total (sin usar solver.Objective())
    2. Cumplimiento de balances y cotas.
    """
    demanda = [40, 60, 75, 25]
    costo_regular = 400
    costo_extra = 450
    trimestres = ["Q1", "Q2", "Q3", "Q4"]

    prod_regular = solucion_df["prod_regular"].values
    prod_extra = solucion_df["prod_extra"].values
    inventario = solucion_df["inventario"].values

    # Objetivo reconstruido: inventario se cobra en los 4 trimestres (como en el modelo)
    costo_total = (
        sum(prod_regular) * costo_regular
        + sum(prod_extra) * costo_extra
        + sum(inventario) * costo_inventario
    )
    print(f"Costo total recalculado: {costo_total:.2f}")

    inventario_prev = inventario_inicial
    passed = True

    for t in range(4):
        prod = prod_regular[t] + prod_extra[t]
        inv_esperado = inventario_prev + prod - demanda[t]
        inv = inventario[t]

        if abs(inv - inv_esperado) > 1e-3:
            print(
                f"Balance inventario no válido en {trimestres[t]}: "
                f"Inventario calculado={inv}, esperado={inv_esperado}"
            )
            passed = False

        if prod_regular[t] < -1e-6 or prod_extra[t] < -1e-6 or inv < -1e-6:
            print(f"Valor negativo detectado en {trimestres[t]}")
            passed = False

        if prod_regular[t] > 40 + 1e-6:
            print(f"prod_regular supera cota 40 en {trimestres[t]}: {prod_regular[t]}")
            passed = False

        if limite_extra is not None and prod_extra[t] > limite_extra + 1e-6:
            print(
                f"prod_extra supera límite {limite_extra} en {trimestres[t]}: {prod_extra[t]}"
            )
            passed = False

        inventario_prev = inv

    if passed:
        print("Todos los balances y restricciones validados correctamente.")
    else:
        print("Algunas restricciones no se cumplen.")

    return passed
```

```python
# Auditoría del caso con límite extra 20 y costo inventario 60
validar_solucion(solucion_df, costo_inventario=60, limite_extra=20)
```

```python
# Candidatos factibles en bordes (variables en cota inferior o superior)
# La solución óptima de LP está en un vértice: varias variables saturadas.

LIMITE_REGULAR = 40
LIMITE_EXTRA = 20  # caso con horas extra acotadas
TOL = 1e-6

candidatos = []
for q in solucion_df.index:
    r = solucion_df.loc[q, "prod_regular"]
    e = solucion_df.loc[q, "prod_extra"]
    i = solucion_df.loc[q, "inventario"]

    if abs(r - 0) <= TOL:
        candidatos.append((q, "prod_regular", r, "cota inferior (0)"))
    if abs(r - LIMITE_REGULAR) <= TOL:
        candidatos.append((q, "prod_regular", r, f"cota superior ({LIMITE_REGULAR})"))

    if abs(e - 0) <= TOL:
        candidatos.append((q, "prod_extra", e, "cota inferior (0)"))
    if abs(e - LIMITE_EXTRA) <= TOL:
        candidatos.append((q, "prod_extra", e, f"cota superior ({LIMITE_EXTRA})"))

    if abs(i - 0) <= TOL:
        candidatos.append((q, "inventario", i, "cota inferior (0)"))

bordes_df = pd.DataFrame(candidatos, columns=["trimestre", "variable", "valor", "borde"])
print("Variables en el borde de la región factible (solución óptima):")
display(bordes_df)

# Planes factibles extremos armados a mano (no necesariamente óptimos)
planes_borde = {
    "producir_temprano": {
        # Adelantar stock en Q1–Q2 con extra al tope
        "prod_regular": [40, 40, 40, 25],
        "prod_extra": [20, 20, 5, 0],
        "inventario": [30, 30, 0, 0],
    },
    "extra_minimo_repartido": {
        # Misma solución del solver con límite 20
        "prod_regular": [40, 40, 40, 25],
        "prod_extra": [5, 20, 20, 0],
        "inventario": [15, 15, 0, 0],
    },
}

print("\nCandidatos factibles en bordes (planes extremos):")
for nombre, plan in planes_borde.items():
    df_plan = pd.DataFrame(plan, index=["Q1", "Q2", "Q3", "Q4"])
    print(f"\n--- plan '{nombre}' ---")
    ok = validar_solucion(df_plan, costo_inventario=60, limite_extra=20)
    print(f"  → {'FACTIBLE' if ok else 'NO FACTIBLE'}")
    display(df_plan)
```

```python
# Prueba de bordes: dos escenarios extremos del modelo Sailco

def construir_y_resolver(limite_extra, costo_inventario, inventario_inicial=10, demanda=None):
    """Arma el modelo Sailco con parámetros de borde y lo resuelve."""
    if demanda is None:
        demanda = [40, 60, 75, 25]

    solver = pywraplp.Solver.CreateSolver("GLOP")
    regs, extras, invs = [], [], []

    for t in range(4):
        regs.append(solver.NumVar(0, 40, f"reg_Q{t+1}"))
        extras.append(solver.NumVar(0, limite_extra, f"extra_Q{t+1}"))
        invs.append(solver.NumVar(0, solver.infinity(), f"inv_Q{t+1}"))

    solver.Add(invs[0] == inventario_inicial + regs[0] + extras[0] - demanda[0])
    for t in range(1, 4):
        solver.Add(invs[t] == invs[t - 1] + regs[t] + extras[t] - demanda[t])

    solver.Minimize(
        sum(400 * regs[t] + 450 * extras[t] + costo_inventario * invs[t] for t in range(4))
    )

    status = solver.Solve()
    resultado = {
        "status": status,
        "optimo": status == pywraplp.Solver.OPTIMAL,
        "factible": status in (pywraplp.Solver.OPTIMAL, pywraplp.Solver.FEASIBLE),
        "objetivo": solver.Objective().Value() if status == pywraplp.Solver.OPTIMAL else None,
        "df": None,
    }
    if resultado["optimo"]:
        resultado["df"] = pd.DataFrame({
            "prod_regular": [v.solution_value() for v in regs],
            "prod_extra": [v.solution_value() for v in extras],
            "inventario": [v.solution_value() for v in invs],
        }, index=["Q1", "Q2", "Q3", "Q4"])
    return resultado


def probar_borde(nombre, limite_extra, costo_inventario, inventario_inicial=10, demanda=None):
    """Resuelve el modelo en un extremo de parámetros e informa factibilidad."""
    print(f"\n=== Borde: {nombre} ===")
    print(
        f"  limite_extra={limite_extra}, costo_inv={costo_inventario}, "
        f"inv0={inventario_inicial}"
    )

    res = construir_y_resolver(limite_extra, costo_inventario, inventario_inicial, demanda)

    if res["optimo"]:
        print(f"  Estado: ÓPTIMO | costo = {res['objetivo']:.2f}")
        display(res["df"])
        validar_solucion(
            res["df"],
            costo_inventario=costo_inventario,
            limite_extra=limite_extra,
            inventario_inicial=inventario_inicial,
        )
    elif res["factible"]:
        print("  Estado: FACTIBLE (sin optimalidad reportada)")
    else:
        print("  Estado: INFACTIBLE (como se esperaba en este borde)")

    return res


# Borde 1: sin horas extra → capacidad insuficiente (déficit ≥ 30)
borde_1 = probar_borde(
    nombre="sin horas extra (limite_extra=0)",
    limite_extra=0,
    costo_inventario=20,
)

# Borde 2: caso del taller (extra ≤ 20, inventario a 60)
borde_2 = probar_borde(
    nombre="inventario caro (60) + extra acotado a 20",
    limite_extra=20,
    costo_inventario=60,
)

print("\nResumen de pruebas de borde:")
print(f"  1) sin extra: {'FACTIBLE' if borde_1['factible'] else 'INFACTIBLE'} (esperado: INFACTIBLE)")
print(
    f"  2) extra≤20 e inv=60: {'ÓPTIMO' if borde_2['optimo'] else 'FALLÓ'} "
    f"(costo={borde_2['objetivo']})"
)
```
