## C05E1 — Transporte balanceado

### Situación completa

SaludRed distribuye kits médicos estandarizados desde dos plantas a tres centros regionales durante la próxima semana. La planta P0 puede despachar hasta 35 lotes y P1 hasta 25. Los centros D0, D1 y D2 deben recibir exactamente 20, 25 y 15 lotes, respectivamente. La oferta total de 60 coincide con la demanda total de 60, por lo que no hace falta inventario ni faltante en este caso. Cada variable x_ij representa lotes enviados desde la planta i al centro j y se considera continua porque un lote de planificación agrega muchas unidades físicas divisibles. El costo por lote depende de la ruta: desde P0 es 4, 6 y 8 hacia D0, D1 y D2; desde P1 es 5, 4 y 3. Los costos están en centenas de unidades monetarias y ya incluyen transporte y manipulación. La responsable logística debe elegir los seis flujos para cumplir toda la demanda sin superar ninguna oferta y minimizar el costo semanal. Los balances de destino expresan servicio obligatorio; las restricciones de planta expresan disponibilidad. Esta distinción será esencial cuando más adelante se permitan inventario o faltantes.

### Pregunta de decisión

### Vocabulario operativo

- **Origen:** Planta desde la que parte un lote; P0 o P1.

- 35 y 25 son lotes máximos disponibles en P0 y P1.

- La fila [5,4,3] contiene costos por lote desde P1 hacia D0, D1 y D2.
