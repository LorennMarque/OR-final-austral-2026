## 0. Entorno y datos

El archivo contiene coeficientes por lote y capacidades. Audite unidades y valores faltantes antes de modelar.

```python
from pathlib import Path
import platform
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import linprog

print(platform.python_version())
df = pd.read_csv(Path('caso_procesamiento.csv'))
df
```

```python
# TODO: separe cargas y capacidades; construya c_max, A y b.

cargas = df[df["tipo"] == "carga"]
capacidades = df[df["tipo"] == "capacidad"]

# Objective coefficients: batch, tiempo_real
c_max = cargas["valor"].to_numpy()

# Constraints:
# computo:   2*batch + 1*tiempo_real <= 100
# operacion: 1*batch + 3*tiempo_real <= 90
A = cargas[["computo", "operacion"]].to_numpy().T

# Resource capacities
b = capacidades[["computo", "operacion"]].to_numpy().flatten()
```

```python
# TODO: resuelva con linprog(method='highs') y registre el estado antes de leer x.
res = linprog(
    c=-c_max,
 A_ub=A,
 b_ub=b,
 bounds=(0, None),
 method="highs"
 )
 
print(res.message)
```

```python
res
```

```python
if res.success:
    print("x =", res.x)
    print("maximum =", -res.fun)
```

```python
def verificar(x, A, b, c, tol=1e-7):
    'Devuelva un diccionario con controles de factibilidad y objetivo.'
    # Verificar no negatividad
    if (x < -tol).any():
        print("Solución contiene valores negativos")
        return {"success": False, "message": "Solución contiene valores negativos"}

    # Verificar consumo de recursos
    consumo = A @ x
    if (consumo - b > tol).any():
        print("Consumo de recursos excede la capacidad")
        return {"success": False, "message": "Consumo de recursos excede la capacidad"}

    # Verificar valor objetivo
    valor_objetivo = c @ x
    if valor_objetivo < -tol:
        print("Valor objetivo es negativo")
        return {"success": False, "message": "Valor objetivo es negativo"}

    # Verificar holguras
    holguras = b - consumo
    if (holguras < -tol).any():
        print("Holguras negativas")
        return {"success": False, "message": "Holguras negativas"}

    return {"success": True, "message": "Solución factible"}

# TODO: aplique la función a res.x.
verificar(res.x, A, b, c_max)
```

```python
# Gráfico: fronteras, región factible, función objetivo y óptimo (caso procesamiento)

# x = [x_B, x_R] = [batch, tiempo_real]
# 2 x_B + x_R <= 100  (cómputo)
# x_B + 3 x_R <= 90   (operación)

x_B = np.linspace(0, 55, 400)
frontera_computo = 100 - 2 * x_B
frontera_operacion = (90 - x_B) / 3

plt.figure(figsize=(8, 6))
plt.plot(x_B, frontera_computo, label='Cómputo: $2x_B + x_R \\leq 100$', color='blue')
plt.plot(x_B, frontera_operacion, label='Operación: $x_B + 3x_R \\leq 90$', color='red')

x_R_top = np.maximum(np.minimum(frontera_computo, frontera_operacion), 0)
plt.fill_between(x_B, 0, x_R_top, where=(x_R_top >= 0), color='gray', alpha=0.3, label='Región factible')

# Agregar la función objetivo como línea de isobeneficio
# c_max debe estar definido, es el vector coeficiente de la función objetivo
# Si la función objetivo es max c_B x_B + c_R x_R, se muestran varias líneas de igual beneficio

# Línea de nivel óptima: 55 x_B + 90 x_R = Z*
Z_opt = float(-res.fun) if res.success else float(c_max @ res.x)
x_R_obj = (Z_opt - c_max[0] * x_B) / c_max[1]
plt.plot(x_B, x_R_obj, '-', label=f'$Z={Z_opt:.0f}$', alpha=0.7)

xB_opt, xR_opt = res.x
plt.plot([xB_opt], [xR_opt], 'ko', label='Solución óptima')
plt.text(xB_opt + 0.8, xR_opt + 0.5, f'({xB_opt:.1f}, {xR_opt:.1f})')

plt.xlim(0, 55)
plt.ylim(0, 35)
plt.xlabel('$x_B$ (lotes batch)')
plt.ylabel('$x_R$ (lotes tiempo real)')
plt.title('Conjunto factible, función objetivo y solución óptima — caso procesamiento')
plt.legend()
plt.grid(True)
plt.show()
```
